```python
from pathlib import Path
from os import listdir
from os import makedirs as mkdir
import json
import pandas as pd
import shutil
```

```python
root = Path.cwd().parent.parent
```

```python
target_file = ['family_summary.csv', 
            'generalization_summary.csv', 
            'output_route_summary_by_category.csv', 
            'per_state_metrics.csv', 
            'per_state_output_metrics.csv', 
            'refinement_depth_ablation.csv', 
            'refinement_iteration_metrics.csv',  
            'state_summary.csv', 
            'test_results.csv', 
            'topology_metrics.csv', 
            'topology_summary_by_category.csv', 
            'training_history.csv']
```

```python
output_result = root / "outputs" / listdir(root / "outputs")[0]
```

```python
# Criar pastas: 1 - pasta results
mkdir(root / 'results', exist_ok=True)
results = root / 'results'
```

```python
# Criar pastas: 2 - Internas results
# 2 - 01_metrics
mkdir(results / '01_metrics', exist_ok=True)
metrics = results / '01_metrics'
# 3 - 02_figures
mkdir(results / '02_figures', exist_ok=True)

# 4 - 03_spectral_analysis
mkdir(results / '03_spectral_analysis', exist_ok=True)
```

```python
solution_json = output_result / "solution_autoencoder_history.json"
manifest_json = output_result / "split_manifest.json"

with solution_json.open("r", encoding="utf-8") as f:
    solution_autoencoder_history = json.load(f)

with manifest_json.open("r", encoding="utf-8") as f:
    split_manifest = json.load(f)


train_count = len(split_manifest["train"])
updates_per_epoch = (train_count + 2 - 1) // 2
df = pd.DataFrame({'stage': 'B_solution_autoencoder', 
                   'epoch': [r['epoch'] for r in solution_autoencoder_history], 
                   'train_loss': [r['training_loss_mean'] for r in solution_autoencoder_history], 
                   'optimizer_update': [(r['epoch'] + 1) * updates_per_epoch for r in solution_autoencoder_history]})
df.to_csv(metrics / 'solution_autoencoder_history_recovered.csv', index=False)
df
```

```python
for f1 in listdir(output_result):
    for f2 in target_file:
        if f1 == f2:
            file = output_result / f1
            shutil.copy(file, metrics)
            print(f1, 'Copy for', metrics)
```
